```python

# Simple Rule-Based Chatbot

print("🤖 ChatBot: Hello! I am your chatbot.")
print("Type 'exit' or 'bye' to end the conversation.\n")

while True:
    user = input("You: ").lower()

    # Exit commands
    if user == "exit" or user == "bye":
        print("🤖 ChatBot: Goodbye! Have a great day.")
        break

    # Greetings
    elif user == "hello" or user == "hi" or user == "hey":
        print("🤖 ChatBot: Hello! How can I help you?")

    # Asking chatbot's name
    elif "name" in user:
        print("🤖 ChatBot: My name is RuleBot.")

    # Asking about health
    elif "how are you" in user:
        print("🤖 ChatBot: I'm doing great! Thanks for asking.")

    # Asking time
    elif "time" in user:
        print("🤖 ChatBot: Sorry, I can't tell the current time yet.")

    # Asking about Python
    elif "python" in user:
        print("🤖 ChatBot: Python is a popular programming language used in AI, web development, and automation.")

    # Thank you
    elif "thank" in user:
        print("🤖 ChatBot: You're welcome!")

    # Default response
    else:
        print("🤖 ChatBot: Sorry, I don't understand that. Please try something else.")
```
